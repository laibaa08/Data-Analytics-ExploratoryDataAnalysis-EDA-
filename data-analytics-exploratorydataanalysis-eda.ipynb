```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

df = sns.load_dataset('titanic')
df.head()
```

```python
print(df.shape)         
print(df.info())       
print(df.describe())    # statistics

# Missing values count
print(df.isnull().sum())

# Visualize missing values
plt.figure(figsize=(10,6))
sns.heatmap(df.isnull(), cbar=False, cmap="viridis")
plt.title("Missing Values Heatmap")
plt.show()
```

```python
# Replace infinite values with NaN
df = df.replace([np.inf, -np.inf], np.nan)

```

```python
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

```

```python
import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

# Countplot for Sex
sns.countplot(x="sex", data=df)
plt.title("Count of Passengers by Sex")
plt.show()

# Countplot for Class
sns.countplot(x="class", data=df)
plt.title("Passenger Count by Class")
plt.show()
```

```python
# Survival by class
sns.countplot(x="class", hue="survived", data=df)
plt.title("Survival Count by Class")
plt.show()

# Survival by sex
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Survival Count by Sex")
plt.show()

# Age vs Survival
plt.figure(figsize=(8,5))
sns.boxplot(x="survived", y="age", data=df)
plt.title("Age vs Survival")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
print(" Key Insights:")
print("- More males traveled, but females had higher survival rates.")
print("- 1st class passengers had better survival chances compared to 3rd class.")
print("- Younger passengers (children) had relatively higher survival rates.")
print("- Fare and class are correlated; higher fare -> better class -> higher survival.")
```
